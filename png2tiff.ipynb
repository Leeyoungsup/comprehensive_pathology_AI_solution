```python
import cv2
from PIL import Image
from glob import glob
import os
import shutil
from tqdm import tqdm
```

```python
image_list=glob('../../project/SS_GA_CPS_annotation/pdl1_main/*.png')
for i in tqdm(range(len(image_list))):
    img=Image.open(image_list[i])
    tiff_image=img.convert('RGB')
    save_path=image_list[i].replace('png','tiff')
    tiff_image.save(save_path)
    os.remove(image_list[i])
    print(f'Saved: {save_path}')
```
